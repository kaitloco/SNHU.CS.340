```python
# import CRUD class 
from CRUD_Python_Module import AnimalShelter
```

```python
# Instantiate class 
shelter = AnimalShelter()
```

```python
# clean test data 
delete_query = {"animal_id": 1} 
shelter.delete(delete_query)
```

```python
# create sample document 
sample_data = { 
    "animal_id": 1, 
    "name": "Buddy", 
    "species": "Dog", 
    "outcome_type": "Adoption" 
}
```

```python
# Test create method 
create_success = shelter.create(sample_data) 
print(f"Insert successful: {create_success}")
```

```python
# Test read method 
query = {"species": "Dog"} 
results = shelter.read(query) 
print("Retrieved documents:", results)
```

```python
# Test update method 
update_query = {"animal_id": 1}
update_data = { 
    "outcome_type": "Transfer" 
} 
updated_records = shelter.update( 
    update_query, 
    update_data 
) 
print(f"Records updated: {updated_records}")
```

```python
# Verify update 
updated_results = shelter.read( 
    {"animal_id": 1} 
) 
print("Updated document:", updated_results)
```

```python
# Test delete method 
delete_query = {"animal_id": 1} 
deleted_records = shelter.delete( 
    delete_query 
) 
print(f"Records deleted: {deleted_records}")
```

```python
# Verify delete 
verify_delete = shelter.read( 
    {"animal_id": 1} 
) 
print("Remaining documents:", verify_delete)
```

```python
# Test invalid read query 
try: 
    shelter.read("Dog") 
    except ValueError as e: 
        print(e)
```

```python
# Test invalid create data 
try: 
    shelter.create(None) 
    except ValueError as e: 
        print(e)
```
